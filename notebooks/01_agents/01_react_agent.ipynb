## How ReAct Works

ReAct (Reasoning + Acting) is a prompting paradigm where the LLM:

1. **Thinks** — Reasons about what to do next
2. **Acts** — Calls a tool (search, calculator, etc.)
3. **Observes** — Sees the tool's output
4. **Repeats** — Until it has enough info to answer

```
User: What's the population of Tokyo times 2?

Think: I need to find Tokyo's population, then multiply by 2.
Act: search("population of Tokyo")
Observe: Tokyo has a population of ~14 million.
Think: Now I multiply 14 million by 2.
Act: calculate("14000000 * 2")
Observe: 28000000
Think: I have the answer.
Answer: 28 million
```

## Key Takeaways

1. **ReAct agents interleave reasoning and action** — they think before acting
2. **Tools extend LLM capabilities** — LLMs can't do math well, but a calculator tool can
3. **LangGraph provides the orchestration** — it manages the state machine (think → act → observe)
4. **Ollama provides local LLM inference** — no API keys needed!
