````python
from agentexplorr.prompt_engineering import PromptTemplate, FewShotSelector
from agentexplorr.prompt_engineering.few_shot import Example

# Create a simple template
template = PromptTemplate(
    "You are a {{ role }}.\n\nTask: {{ task }}\n\nRespond in {{ format }} format.",
    name="simple_system"
)

prompt = template.render(
    role="expert Python developer",
    task="Explain list comprehensions",
    format="markdown"
)

print(f"Template variables: {template.variables}")
print(f"\nRendered prompt:\n{prompt}")
````

````python
# Few-shot learning — teach by example
selector = FewShotSelector()

selector.add_examples([
    Example("I love this product!", "Positive"),
    Example("Terrible experience, never again.", "Negative"),
    Example("It was okay, nothing special.", "Neutral"),
    Example("Best purchase I've ever made!", "Positive"),
    Example("Broke after one day.", "Negative"),
])

prompt = selector.build_prompt(
    instruction="Classify the sentiment of the following text.",
    input_text="Pretty decent for the price.",
    n=3,
)

print(prompt)
````

````python
# Chain of Thought prompting
from agentexplorr.prompt_engineering import ChainOfThoughtPrompt
from agentexplorr.prompt_engineering.chain_of_thought import ReasoningStep

cot = ChainOfThoughtPrompt(task="Solve the following math problem.")

# Add an example with explicit reasoning
cot.add_example(
    question="What is 15% of 80?",
    reasoning=[
        ReasoningStep(1, "Convert to decimal", "15% = 15/100 = 0.15"),
        ReasoningStep(2, "Multiply", "0.15 × 80 = 12"),
    ],
    answer="12"
)

prompt = cot.build_few_shot("What is 20% of 150?")
print(prompt)
````

````python
# Structured output parsing
from pydantic import BaseModel
from agentexplorr.prompt_engineering import OutputParser

class MovieReview(BaseModel):
    title: str
    rating: float
    sentiment: str
    summary: str

parser = OutputParser()

# Simulate LLM output (in practice, this comes from the model)
llm_output = '''
Here's my analysis:
```json
{"title": "Inception", "rating": 9.2, "sentiment": "positive", "summary": "A mind-bending masterpiece"}
```
'''

review = parser.parse_json_to_model(llm_output, MovieReview)
print(f"Title: {review.title}")
print(f"Rating: {review.rating}")
print(f"Sentiment: {review.sentiment}")
print(f"Summary: {review.summary}")

# Get format instructions to include in your prompt
instructions = OutputParser.get_format_instructions(MovieReview)
print(f"\nFormat instructions for LLM:\n{instructions}")
````

## Key Takeaways

1. **Templates make prompts reusable and version-controlled**
2. **Few-shot examples teach the model by showing, not telling**
3. **Chain of Thought unlocks reasoning** — just "think step by step"
4. **Structured output turns LLM text into validated Python objects**
