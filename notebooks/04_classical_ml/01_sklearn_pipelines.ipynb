# Scikit-Learn Pipelines — Classification, Regression, Clustering

## What You'll Learn

- Regression with Ridge/Lasso (California Housing)
- Clustering with KMeans (Iris)
- Hyperparameter tuning with GridSearchCV

```python
from agentexplorr.classical_ml.pipelines.classification import ClassificationPipeline

# Create and train a classification pipeline on the Wine dataset
clf = ClassificationPipeline()
metrics = clf.train_on_wine()

print("Classification Results (Wine Dataset):")
for metric, value in metrics.items():
    if isinstance(value, float):
        print(f"  {metric}: {value:.4f}")
```

```python
from agentexplorr.classical_ml.pipelines.regression import RegressionPipeline

# Train a regression pipeline on California Housing
reg = RegressionPipeline()
metrics = reg.train_on_housing(max_samples=2000)

print("Regression Results (California Housing):")
for metric, value in metrics.items():
    if isinstance(value, float):
        print(f"  {metric}: {value:.4f}")
```

```python
from agentexplorr.classical_ml.pipelines.clustering import ClusteringPipeline

# Cluster the Iris dataset
cluster = ClusteringPipeline(n_clusters=3)
metrics = cluster.fit_on_iris()

print("Clustering Results (Iris):")
for metric, value in metrics.items():
    if isinstance(value, float):
        print(f"  {metric}: {value:.4f}")
```

## Key Takeaways

1. **Pipelines prevent data leakage** — preprocessing is fit only on training data
2. **GridSearchCV automates hyperparameter tuning** with cross-validation
3. **Silhouette score measures cluster quality** (higher = better separated clusters)
4. **R² score measures regression quality** (1.0 = perfect, 0.0 = no better than mean)
