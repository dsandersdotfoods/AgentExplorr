## The Alpaca Format

The Alpaca dataset uses a simple instruction-following format:

```
### Instruction:
{instruction}

### Input:
{optional input context}

### Response:
{expected output}
```

This format teaches the model to:
1. Read an instruction
2. Optionally use additional context
3. Generate a helpful response

## Key Takeaways

1. **Data quality > data quantity** — a well-formatted small dataset beats a messy large one
2. **Consistent formatting is critical** — the model learns the pattern from the format
3. **Always hold out a validation set** — to detect overfitting
4. **Tokenization matters** — padding and truncation affect training efficiency
